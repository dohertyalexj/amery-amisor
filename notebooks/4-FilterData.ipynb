## Apply Filtering

### 30d Filtering

Remove tidal, short- to medium-term influences and constrain a seasonal cycle.

```python
# Define the cutoff period for filtering in days (or hours)
cutoff_period_days = 30 # days
cutoff_period = cutoff_period_days*24 # hours

# The sampling interval is 0.5 hours, given that each MicroCAT records observations every 30 minutes
sampling_interval=0.5

# Define a period of time in days for less than which we are comfortable interpolating for 
# Set this to 0 for no interpolation
interpolate_days = 7.5 # days
max_gap_hours=interpolate_days*24 # hours

# Apply filtering to the DataFrame
# NOTE: cutoff_period is defined in terms of hours
# Iterate over each unique combination of borehole and category in the DataFrame
for (bh, cat), group in df.groupby(["borehole", "cat"]):
    # Create a boolean mask to identify the rows corresponding to the current iteration of borehole and cat
    mask = (df["borehole"] == bh) & (df["cat"] == cat)
    # Apply the low-pass filter to the "in_situ_temperature" column 
    # Assign the results back to a new "in_situ_temperature_smooth" column 
    df.loc[mask, "in_situ_temperature_smooth_30d"] = lowpass_filter(
        group["in_situ_temperature"].values,
        cutoff_period=cutoff_period,
        sampling_interval=sampling_interval,
        max_gap_hours=max_gap_hours
    )
    df.loc[mask, "in_situ_freezing_point_temperature_smooth_30d"] = lowpass_filter(
        group["in_situ_freezing_point_temperature"].values,
        cutoff_period=cutoff_period,
        sampling_interval=sampling_interval,
        max_gap_hours=max_gap_hours
    )
    df.loc[mask, "potential_temperature_smooth_30d"] = lowpass_filter(
        group["potential_temperature"].values,
        cutoff_period=cutoff_period,
        sampling_interval=sampling_interval,
        max_gap_hours=max_gap_hours
    )
    df.loc[mask, "conservative_temperature_smooth_30d"] = lowpass_filter(
        group["conservative_temperature"].values,
        cutoff_period=cutoff_period,
        sampling_interval=sampling_interval,
        max_gap_hours=max_gap_hours
    )
    df.loc[mask, "conservative_freezing_point_temperature_smooth_30d"] = lowpass_filter(
        group["conservative_freezing_point_temperature"].values,
        cutoff_period=cutoff_period,
        sampling_interval=sampling_interval,
        max_gap_hours=max_gap_hours
    )
    df.loc[mask, "practical_salinity_smooth_30d"] = lowpass_filter(
        group["practical_salinity"].values,
        cutoff_period=cutoff_period,
        sampling_interval=sampling_interval,
        max_gap_hours=max_gap_hours
    )
    df.loc[mask, "absolute_salinity_smooth_30d"] = lowpass_filter(
        group["absolute_salinity"].values,
        cutoff_period=cutoff_period,
        sampling_interval=sampling_interval,
        max_gap_hours=max_gap_hours
    )
```
